```python
# f-string é o nome dado a este f antes das(a) aspa(s) que permite a formatação do texto e o uso para calculos e numeros usando {} sem a necessidade de fechar as aspas e por virgula para separar.
# :.2f serve para escolher quantidade de casas decimais. basta substituir o 2 pela quantidade de casas que desejar.
print(f'''imc: {75.3 / 1.83 ** 2:.2f}''')
```
