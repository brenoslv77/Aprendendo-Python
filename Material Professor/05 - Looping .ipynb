### Exercício: Jogo de Adivinhação

Crie um programa em Python que sorteie um número aleatório entre 1 e 10 utilizando a biblioteca `random`.

O usuário deverá tentar adivinhar o número sorteado, tendo **no máximo 5 tentativas**.

O programa deve:

1. Utilizar `random` para sortear um número entre 0 e 10.
2. Utilizar um laço `for` para controlar as 5 tentativas.
3. A cada tentativa, solicitar que o usuário digite um número.
6. Se o usuário não acertar em nenhuma das 5 tentativas, informar qual era o número sorteado.

````python
from random import randint

num_pc = randint(0,10)
# print(num_pc)
for j in range(5):
    usr = int(input('Informe um valor de 0 a 10: '))
    if usr == num_pc:
        print('Acertou 👌')
        #break - para o looping ATUAL
        break
    else:
        print('Errou 😒')
print('O numero era', num_pc)
````

### Exercício: Jogo de Adivinhação

Crie um programa em Python que sorteie um número aleatório entre 1 e 100 utilizando a biblioteca `random`.

O usuário deverá tentar adivinhar o número sorteado, tendo **no máximo 5 tentativas**.

O programa deve:

1. Utilizar `random` para sortear um número entre 1 e 100.
2. Utilizar um laço `for` para controlar as 5 tentativas.
3. A cada tentativa, solicitar que o usuário digite um número.
4. Informar se o número digitado é maior ou menor que o número sorteado.
5. Caso o usuário acerte, exibir uma mensagem de parabéns e encerrar as tentativas.
6. Se o usuário não acertar em nenhuma das 5 tentativas, informar qual era o número sorteado.

**Desafio:** tente utilizar `break` para encerrar o laço assim que o usuário acertar o número.

````python
from random import randint
from time import sleep
from google.colab import output
#sleep(segundos) - para o programa por alguns segundos
#output.clear() - limpa a saida do programa

num_pc = randint(0,100)
# print(num_pc)
for j in range(5):
    usr = int(input('Informe um valor de 0 a 100: '))
    if usr == num_pc:
        print('Acertou 👌')
        #break - para o looping ATUAL
        break
    else:
        print('Errou 😒')
        if num_pc > usr:
            print(f'⬆️ O número sorteado é maior ⬆️')
        else:
            print(f'⬇️ O número sorteado é menor ⬇️')
        sleep(3)
        output.clear()
print('O numero era', num_pc)
````

 Maior Número e Quantidade de Números Digitados
Escreva um programa em Python que solicite ao usuário a digitação de vários números. O programa deverá continuar pedindo números até que o usuário digite 0, que será utilizado para encerrar a entrada de dados.

Ao final, o programa deverá:

Identificar e exibir o maior número digitado.

Informar a quantidade de números digitados, desconsiderando o zero utilizado para encerrar o programa.

Utilize uma estrutura de repetição while para realizar a leitura dos números e estruturas condicionais para verificar qual é o maior número.

````python
maior_numero = float('-inf')
while True:
    num = float(input('Digite um número: "0 para sair" '))
    if num == 0:
        break
    elif num > maior_numero:
        maior_numero = num
print(f'O maior número digitado foi {maior_numero}')
````


### Exercício: Jogo de Adivinhação com `while`

Crie um programa em Python que sorteie um número aleatório entre 1 e 20 utilizando a biblioteca `random`.

O usuário deverá tentar adivinhar o número sorteado, e o programa continuará solicitando novos palpites até que ele acerte.

O programa deve:

1. Sortear um número aleatório entre 1 e 20 utilizando `random`.
2. Utilizar um laço `while` para repetir as tentativas até o usuário acertar.
3. Solicitar que o usuário digite um número a cada tentativa.
4. Informar se o número digitado é maior ou menor que o número sorteado.
5. Contar quantas tentativas o usuário realizou.
6. Quando o usuário acertar, exibir uma mensagem de parabéns e informar o número total de tentativas realizadas.

**Desafio:** inicialize um contador antes do laço `while` e atualize seu valor a cada palpite.

````python
from random import randint
from time import sleep
from google.colab import output

contador = 0
num_pc = randint(1,20)

while True:
    print(f'Tentativas: {contador*'💔'}')
    num = int(input('Informe um número entre 1 e 20: '))
    if num == num_pc:
        print('Você venceu!')
        break
    sleep(2.5)
    output.clear()
    contador+=1
````
