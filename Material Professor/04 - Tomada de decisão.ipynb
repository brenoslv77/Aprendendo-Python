```python
altura = float(input('Informe a altura: '))
peso = float(input('Informe o peso: '))
imc = peso/altura**2
if imc < 18.5:
    print('Abaixo do peso!')
elif imc < 25:
    print('Peso normal!')
elif imc < 30:
    print('Excesso de peso!')
elif imc < 35:
    print('Obesidade 1!')
elif imc < 40:
    print('Obesidade 2!')
else:
    print('Obesidade 3!')
print(f'IMC: {imc:.2f}')
```

Desconto na loja:

Crie um programa que peça o valor de uma compra.

O programa deve informar que o cliente ganhou um desconto especial se a compra for maior que R$ 200 OU se o cliente possuir um cupom de desconto.

```python
valor_compra = float(input('Informe o valor da compra: '))
cupom = input('Possui cupom? (s/n) \n')
if cupom == 's' or valor_compra > 200:
    print(f'Total a pagar: R${valor_compra*0.90:.2f}')
else:
    print(f'Total a pagar: R${valor_compra}')
```

Idade:

- Solicite o ano de nascimento.

- Calcule a idade considerando o ano atual (use a função datetime.now()
do módulo datetime).

- Exiba a idade.

```python
from datetime import datetime

#.year pega o ano
#.month pega mes
#.day pega dia

ano_atual = datetime.now().year
ano_nasc = int(input('Informe o ano de nascimento: '))
idade = ano_atual - ano_nasc
print(f'Você tem {idade} anos')
```
